```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# -------------------------------------------
# CONFIG
# -------------------------------------------

base_path = r'C:/Users/Abel/Downloads/PAN4/Semester 1/Data Analytics and Data Mining (CSC1171)/group_project/new_datasets'

datasets = {
    "forbes2000_monthly":  f"{base_path}/forbes2000_monthly.csv",
    "forbes2000_yearly":   f"{base_path}/forbes2000_yearly.csv",
    "sp500_monthly":       f"{base_path}/sp500_monthly.csv",
    "sp500_yearly":        f"{base_path}/sp500_yearly.csv",
    "nasdaq_monthly":      f"{base_path}/nasdaq_monthly.csv",
    "nasdaq_yearly":       f"{base_path}/nasdaq_yearly.csv",
    "nyse_monthly":        f"{base_path}/nyse_monthly.csv",
    "nyse_yearly":         f"{base_path}/nyse_yearly.csv",
}

numeric_features = [
    'avg_return',
    'volatility',
    'avg_adj_close',
    'avg_volume',
    'avg_range_pct'
]

# -------------------------------------------
# HELPER: low-correlation feature selection
# -------------------------------------------

def select_low_correlation_features(corr_matrix, threshold=0.3):
    """Select features with correlation below threshold, based on average abs correlation."""
    corr_abs = corr_matrix.abs()
    avg_corr = corr_abs.mean()

    selected_features = []
    # start with feature that has lowest average correlation
    first = avg_corr.idxmin()
    selected_features.append(first)

    # add features that stay below threshold w.r.t. all already selected
    for feature in avg_corr.sort_values().index:
        if feature in selected_features:
            continue
        correlations = [abs(corr_matrix.loc[feature, sel]) for sel in selected_features]
        if all(c < threshold for c in correlations):
            selected_features.append(feature)

    return selected_features, avg_corr.sort_values()

# -------------------------------------------
# MAIN ANALYSIS FUNCTION FOR ONE DATASET
# -------------------------------------------

def analyse_dataset(df, name):
    print("\n" + "="*80)
    print(f"DATASET: {name}")
    print("="*80)

    # -----------------------
    # 0. Prep data
    # -----------------------
    # Only keep rows where all numeric features exist
    X = df[numeric_features].copy()
    X = X.fillna(X.mean())

    # Standardise
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # -----------------------
    # 1. Correlation analysis
    # -----------------------
    print("\n" + "="*60)
    print("CORRELATION MATRIX")
    print("="*60)
    corr_matrix = X.corr()
    print(corr_matrix)

    plt.figure(figsize=(8, 6))
    sns.heatmap(
        corr_matrix, annot=True, cmap='coolwarm',
        center=0, fmt='.2f', square=True
    )
    plt.title(f'Feature Correlation Matrix ({name})')
    plt.tight_layout()
    plt.show()

    # Least correlated pairs
    print("\n" + "="*60)
    print("LEAST CORRELATED FEATURE PAIRS")
    print("="*60)
    corr_abs = corr_matrix.abs()
    np.fill_diagonal(corr_abs.values, np.nan)
    unstacked = corr_abs.unstack()
    sorted_corr = unstacked.sort_values(ascending=True)
    print(sorted_corr.head(10))

    # -----------------------
    # 2. PCA
    # -----------------------
    print("\n" + "="*60)
    print("PCA ANALYSIS")
    print("="*60)

    pca = PCA()
    X_pca = pca.fit_transform(X_scaled)

    explained_variance = pca.explained_variance_ratio_
    cumulative_variance = np.cumsum(explained_variance)

    print("\nExplained Variance by Component:")
    for i, var in enumerate(explained_variance):
        print(f"PC{i+1}: {var:.4f} ({var*100:.2f}%)")
    print(f"\nCumulative Variance: {cumulative_variance}")

    # Scree + cumulative plots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    ax1.bar(range(1, len(explained_variance)+1), explained_variance, color='black')
    ax1.set_xlabel('Principal Component')
    ax1.set_ylabel('Variance Ratio')
    ax1.set_title(f'Scree Plot ({name})')
    ax1.set_xticks(range(1, len(explained_variance)+1))

    ax2.plot(range(1, len(cumulative_variance)+1), cumulative_variance, 'bo-')
    ax2.axhline(y=0.95, color='r', linestyle='--', label='95% variance')
    ax2.set_xlabel('Number of Components')
    ax2.set_ylabel('Cumulative Explained Variance')
    ax2.set_title(f'Cumulative Explained Variance ({name})')
    ax2.legend()
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # -----------------------
    # 3. PCA loadings
    # -----------------------
    print("\n" + "="*60)
    print("PCA LOADINGS (Feature Contributions)")
    print("="*60)

    loadings = pd.DataFrame(
        pca.components_.T,
        columns=[f'PC{i+1}' for i in range(len(numeric_features))],
        index=numeric_features
    )
    print(loadings)

    plt.figure(figsize=(10, 4))
    sns.heatmap(
        loadings, annot=True, cmap='RdBu_r', center=0,
        fmt='.2f', cbar_kws={'label': 'Loading'}
    )
    plt.title(f'PCA Loadings ({name})')
    plt.xlabel('Principal Components')
    plt.ylabel('Features')
    plt.tight_layout()
    plt.show()

    # -----------------------
    # 4. Biplot for PC1–PC2
    # -----------------------
    fig, ax = plt.subplots(figsize=(7, 6))

    scatter = ax.scatter(
        X_pca[:, 0], X_pca[:, 1],
        alpha=0.4, s=10, c=range(len(X_pca)), cmap='viridis'
    )
    ax.set_xlabel(f'PC1 ({explained_variance[0]*100:.1f}%)')
    ax.set_ylabel(f'PC2 ({explained_variance[1]*100:.1f}%)')
    ax.set_title(f'PCA Biplot: PC1 vs PC2 ({name})')

    # loading vectors
    for i, feature in enumerate(numeric_features):
        ax.arrow(
            0, 0,
            pca.components_[0, i]*3,
            pca.components_[1, i]*3,
            head_width=0.05, head_length=0.05,
            fc='red', ec='red', alpha=0.7
        )
        ax.text(
            pca.components_[0, i]*3.3,
            pca.components_[1, i]*3.3,
            feature,
            fontsize=8, ha='center',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.6)
        )

    ax.grid(True, alpha=0.3)
    plt.colorbar(scatter, label='Sample Index')
    plt.tight_layout()
    plt.show()

    # -----------------------
    # 5. Recommended features
    # -----------------------
    print("\n" + "="*60)
    print("RECOMMENDED FEATURES (Least Correlated)")
    print("="*60)

    low_corr_features, avg_corr_sorted = select_low_correlation_features(corr_matrix, threshold=0.3)
    print("\nFeatures with correlation < 0.3 w.r.t. selected set:")
    print(low_corr_features)

    print("\n" + "="*60)
    print("AVERAGE ABSOLUTE CORRELATION FOR EACH FEATURE")
    print("="*60)
    print(avg_corr_sorted)

    print("\n✓ Features with lowest average correlation:")
    print(avg_corr_sorted.head(3).index.tolist())


# -------------------------------------------
# RUN FOR ALL 8 DATASETS
# -------------------------------------------

for name, path in datasets.items():
    df = pd.read_csv(path)
    analyse_dataset(df, name)
```
